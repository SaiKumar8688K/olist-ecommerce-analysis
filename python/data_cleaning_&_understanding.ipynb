```python
avg_item_price = (
    sales_product[
        sales_product['year_month'].isin([
            pd.Period('2017-11', freq='M'),
            pd.Period('2018-05', freq='M')
        ])
    ]
    .groupby(['year_month', 'product_category_name'])
    .agg(
        revenue = ('price','sum'),
        total_items = ('order_item_id','count')
    )
    .reset_index()
)
avg_item_price['average_item_price'] = (
    avg_item_price['revenue'] / avg_item_price['total_items']
)
```

```python
avg_item_price
```

```python
category_comparison = (
    avg_item_price
    .pivot(
        index='product_category_name',
        columns='year_month',
        values='average_item_price'
    )
    .reset_index()
)

category_comparison['price_change'] = (
    category_comparison[pd.Period('2018-05', freq='M')]
    - category_comparison[pd.Period('2017-11', freq='M')]
)

category_comparison.sort_values(
    'price_change',
    ascending=False
).head(10)
```
